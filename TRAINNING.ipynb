**Reasoning**:
The previous code failed due to a floating-point precision issue when checking if the split ratios sum to 1.0. I will fix this by calculating the `test_ratio` dynamically from the remaining percentage to ensure the sum is exactly 1.0.



```python
import os
import random
import shutil

# 1. Define the input directory containing the image and annotation files
input_data_dir = "/content/drive/MyDrive/Percobaan_Trainning_Warna/DATA_RAW/Perpaduan_Warna_Campuran"

# 2. Define the output base directory where the split data will be stored
output_base_dir = "/content/drive/MyDrive/Percobaan_Trainning_Warna/DATA_SPLIT"

# 3. Define the desired split ratios for training, validation, and testing sets
train_ratio = 0.7
val_ratio = 0.2
test_ratio = 1.0 - (train_ratio + val_ratio) # Calculate test_ratio to ensure sum is exactly 1.0

# Ensure the ratios sum to 1 (this check is now primarily for verification if individual ratios were set manually)
if not (abs(train_ratio + val_ratio + test_ratio - 1.0) < 1e-9):
    raise ValueError("Split ratios must sum to 1.0")

# 4. Initialize empty lists to store the paths of image and annotation files
image_files = []
annotation_files = []
```

**Reasoning**:
I will iterate through the input directory to identify and group corresponding image and annotation file pairs while excluding 'classes.txt'.



```python
import os

# 5. Iterate through the input directory to identify and group corresponding image and annotation file pairs
#    Exclude 'classes.txt'.

# Dictionary to store base filenames (e.g., 'frame_00000') and their corresponding image and annotation paths
file_pairs = {}

# List all files in the input directory
all_files = os.listdir(input_data_dir)

for file_name in all_files:
    if file_name == "classes.txt":
        continue # Skip classes.txt

    # Get the base name without extension (e.g., 'frame_00000' from 'frame_00000.png' or 'frame_00000.txt')
    base_name, ext = os.path.splitext(file_name)

    if base_name not in file_pairs:
        file_pairs[base_name] = {"image": None, "annotation": None}

    full_path = os.path.join(input_data_dir, file_name)

    if ext.lower() == '.png': # Assuming images are PNG files
        file_pairs[base_name]["image"] = full_path
    elif ext.lower() == '.txt': # Assuming annotations are TXT files
        file_pairs[base_name]["annotation"] = full_path

# Filter out any base names that don't have both an image and an annotation file
# And convert the dictionary values to a list of (image_path, annotation_path) tuples
valid_file_pairs = [
    (info["image"], info["annotation"])
    for base_name, info in file_pairs.items()
    if info["image"] is not None and info["annotation"] is not None
]

print(f"Found {len(valid_file_pairs)} valid image-annotation pairs.")
print("First 5 valid file pairs:")
for i, pair in enumerate(valid_file_pairs):
    if i >= 5: break
    print(f"  Image: {os.path.basename(pair[0])}, Annotation: {os.path.basename(pair[1])}")

# Populate image_files and annotation_files lists for shuffling later
# This step might be redundant if we directly split from valid_file_pairs, but keeping it for consistency with instructions
for img_path, ann_path in valid_file_pairs:
    image_files.append(img_path)
    annotation_files.append(ann_path)
```

**Reasoning**:
Now that the valid image-annotation pairs are identified, I will randomly shuffle this list to ensure an unbiased distribution when splitting the dataset. Then, I will calculate the number of samples for each split (training, validation, and testing) based on the defined ratios.



```python
import random

# 6. Randomly shuffle the list of grouped file pairs
random.shuffle(valid_file_pairs)

# 7. Calculate the number of samples for each split based on the defined ratios
total_samples = len(valid_file_pairs)
train_count = int(total_samples * train_ratio)
val_count = int(total_samples * val_ratio)
test_count = total_samples - train_count - val_count # Ensure test_count covers any remainder due to int conversion

print(f"Total samples: {total_samples}")
print(f"Train samples: {train_count}")
print(f"Validation samples: {val_count}")
print(f"Test samples: {test_count}")

# Split the shuffled list into train, val, and test sets
train_pairs = valid_file_pairs[:train_count]
val_pairs = valid_file_pairs[train_count : train_count + val_count]
test_pairs = valid_file_pairs[train_count + val_count :]
```

**Reasoning**:
I will create the necessary output subdirectories for each split (train, validation, and test) including separate folders for 'images' and 'labels' within each split directory. This sets up the directory structure where the files will be moved.



```python
import os

# 8. Create the necessary output subdirectories
def create_directories(base_dir, splits=["train", "val", "test"]):
    for split in splits:
        os.makedirs(os.path.join(base_dir, split, "images"), exist_ok=True)
        os.makedirs(os.path.join(base_dir, split, "labels"), exist_ok=True)
    print(f"Created directory structure under {base_dir}")

create_directories(output_base_dir)
```

**Reasoning**:
I will iterate through the shuffled train, validation, and test file pairs and move the image files to their respective 'images' subdirectories and annotation files to their 'labels' subdirectories within the `output_base_dir` for each split.



```python
import shutil
import os

# 9. Iterate through the shuffled file pairs and move them to their respective directories
def move_files(file_pairs, split_name, base_output_dir):
    split_images_dir = os.path.join(base_output_dir, split_name, "images")
    split_labels_dir = os.path.join(base_output_dir, split_name, "labels")

    print(f"Moving {len(file_pairs)} files to {split_name}...")
    for img_path, ann_path in file_pairs:
        # Move image file
        shutil.move(img_path, os.path.join(split_images_dir, os.path.basename(img_path)))
        # Move annotation file
        shutil.move(ann_path, os.path.join(split_labels_dir, os.path.basename(ann_path)))
    print(f"Finished moving files for {split_name}.")

# Move files for each split
move_files(train_pairs, "train", output_base_dir)
move_files(val_pairs, "val", output_base_dir)
move_files(test_pairs, "test", output_base_dir)

print("Data splitting and moving complete!")
```

## Apply Augmentations and Save New Data

### Subtask:
Define a set of augmentation transforms (e.g., horizontal flip, small rotations, brightness/contrast changes) using Albumentations. Iterate through the selected subset of data, apply these transforms, and save the newly generated images and their corresponding YOLO format annotation files back into the training directories, ensuring unique filenames.


**Reasoning**:
I will define the augmentation pipeline using `albumentations.Compose` with specified transforms and `bbox_params` for YOLO format. This will prepare the augmentation object to be applied to the image and bounding box data.



```python
import albumentations as A
import cv2

# 1. Define an augmentation pipeline using albumentations.Compose
transform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.2),
    A.ShiftScaleRotate(shift_limit=0.0625, scale_limit=0.1, rotate_limit=15, p=0.5),
    A.HueSaturationValue(hue_shift_limit=20, sat_shift_limit=30, val_shift_limit=20, p=0.2),
], bbox_params=A.BboxParams(format='yolo', label_fields=['class_ids']))

print("Augmentation pipeline defined.")
```

**Reasoning**:
I will proceed with iterating through the selected samples, applying the defined augmentation pipeline, and saving the augmented images and their corresponding YOLO annotation files to the training directories, ensuring unique filenames as per the instructions.



```python
import cv2
import os
import numpy as np

augmented_count = 0

for i, (image_path, annotation_path) in enumerate(samples_for_augmentation):
    # 4. Read the image using cv2.imread()
    image = cv2.imread(image_path)
    if image is None:
        print(f"Warning: Could not read image {image_path}. Skipping.")
        continue

    # 5. Read the annotation file and parse bounding boxes
    bboxes = []
    class_ids = []
    try:
        with open(annotation_path, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) == 5:
                    class_id = int(parts[0])
                    x_center, y_center, width, height = map(float, parts[1:])
                    bboxes.append([x_center, y_center, width, height])
                    class_ids.append(class_id)
    except FileNotFoundError:
        print(f"Warning: Annotation file {annotation_path} not found. Skipping.")
        continue
    except Exception as e:
        print(f"Warning: Error reading annotation file {annotation_path}: {e}. Skipping.")
        continue

    # 6. Apply the defined augmentation pipeline
    # Albumentations expects class_ids as a list for label_fields
    augmented = transform(image=image, bboxes=bboxes, class_ids=class_ids)
    augmented_image = augmented['image']
    augmented_bboxes = augmented['bboxes']
    augmented_class_ids = augmented['class_ids']

    # 7. Generate unique filenames for the augmented image and its corresponding annotation file
    original_base_name = os.path.splitext(os.path.basename(image_path))[0]
    new_image_name = f"{original_base_name}_aug_{augmented_count}.png"
    new_annotation_name = f"{original_base_name}_aug_{augmented_count}.txt"

    new_image_path = os.path.join(train_images_dir, new_image_name)
    new_annotation_path = os.path.join(train_labels_dir, new_annotation_name)

    # 8. Save the augmented image
    cv2.imwrite(new_image_path, augmented_image)

    # 9. Save the augmented bounding boxes and class IDs to a new annotation file
    with open(new_annotation_path, 'w') as f:
        for bbox, class_id in zip(augmented_bboxes, augmented_class_ids):
            # YOLO format: class_id x_center y_center width height
            f.write(f"{class_id} {bbox[0]} {bbox[1]} {bbox[2]} {bbox[3]}\n")

    # 10. Increment the augmented_count
    augmented_count += 1

# 11. After the loop, print the total number of new augmented samples created.
print(f"Successfully created {augmented_count} new augmented samples.")
```
